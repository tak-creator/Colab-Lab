### importing, loading, printing

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv('Data.csv')
x = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

```python
print(x)
```

```python
print(y)
```

### missing data

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
imputer.fit(x[:, 1:3])
x[:, 1:3] = imputer.transform(x[:, 1:3])
```

```python
print(x)
```

### encoding categorical data

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct =ColumnTransformer(
    transformers=[('encoder', OneHotEncoder(), [0])], # te 0 to która kolumna !
    remainder='passthrough' # Keeps the remaining columns as they are. A transformer object: Applies a specific transformer to the remaining columns.
    )
x = np.array(ct.fit_transform(x))
```

```python
print(x)
```

### encoding dependent viariable

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(y)
```

```python
print(y)
```

### splitting data -> train/test sets

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
print(x_train, '\n')
print(x_test, '\n')
print(y_train, '\n')
print(y_test, ' :)')
```

### feature scalling
---
    
    * standaryzacja - przekształcenie cechy tak, by miała średnią 0 i odchylenie standardowe 1 (czyli tzw. rozkład standardowy).

    * normalizacja - sprowadzenie wartości cechy do z góry określonego przedziału, zwykle [0, 1] lub [-1, 1]

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train[:, 3:] = sc.fit_transform(x_train[:, 3:])
x_test[:, 3:] = sc.transform(x_test[:, 3:])
```

```python
print(x_train, '\n')
print(x_test, '\n')
```
