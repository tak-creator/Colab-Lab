```python
import numpy as np
import torch
```

## Transpose

```python
nv = np.array([[1, 2, 3, 4]])
print(nv)

print(nv.T) # transposing

nvt = nv.T
print(nvt.T)
```

```python
nM = np.array([[1, 2, 3, 4],
                [5, 6, 7, 8]
               ])
print(nM)

print(nM.T)

nMt = nM.T
print(nMt.T)
```

```python
tv = torch.tensor([[1, 2, 3, 4]])
print(tv)

print(tv.T) # transposing

tvT = tv.T
print(tvT.T)
```

```python
tM = torch.tensor([[1, 2, 3, 4],
                [5, 6, 7, 8]
               ])
print(tM)

print(tM.T)

tMt = tM.T
print(tMt.T)
```

```python
print(f'Type of nv {type(nv)}')
print(f'Type of nM {type(nM)}')
print(f'Type of tv {type(tv)}')
print(f'Type of tM {type(tM)}')
```

## Dot product

(dwa vectory na każdym miejscu mnoże (np. nr jedned z nim) a potem wymnożenie dodaje)

```python
nv1 = np.array([1, 2, 3, 4])
nv2 = np.array([-5, -6, 7, 2])

print(np.dot(nv1, nv2))
print(np.sum(nv1 * nv2))
```

```python
tv1 = torch.tensor([1, 2, 3, 4])
tv2 = torch.tensor([-5, -6, 7, 2])

print(torch.dot(tv1, tv2))
print(torch.sum(tv1 * tv2))
```
